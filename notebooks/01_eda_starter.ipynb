# Car Price EDA Starter\n\nThis notebook serves as a starting point for exploratory data analysis on car price data.

```python
from pathlib import Path\n\nimport matplotlib.pyplot as plt\nimport pandas as pd\nimport seaborn as sns\n\nsns.set_theme(style='whitegrid')\nplt.rcParams['figure.figsize'] = (10, 6)\n\nPROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()\nDATA_PATH = PROJECT_ROOT / 'data' / 'raw' / 'cars.csv'\nDATA_PATH
```

```python
df = pd.read_csv(DATA_PATH)\ndf.head()
```

```python
df.describe(include='all').T
```

```python
missing = df.isna().sum().sort_values(ascending=False)\nmissing[missing > 0]
```

```python
numeric_cols = df.select_dtypes(include='number').columns\ndf[numeric_cols].hist(bins=30, figsize=(14, 10))\nplt.tight_layout()
```

```python
if 'price' in df.columns:\n    sns.histplot(df['price'].dropna(), kde=True)\n    plt.title('Price distribution')\n    plt.show()
```

```python
if {'price', 'mileage'}.issubset(df.columns):\n    sns.scatterplot(data=df, x='mileage', y='price')\n    plt.title('Mileage vs price')\n    plt.show()
```

```python
if {'price', 'brand'}.issubset(df.columns):\n    top_brands = df['brand'].value_counts().head(10).index\n    plot_df = df[df['brand'].isin(top_brands)].copy()\n    sns.boxplot(data=plot_df, x='brand', y='price')\n    plt.xticks(rotation=45)\n    plt.title('Price by top brands')\n    plt.show()
```
